```python
from pyspark.sql.functions import *

#========================================================
#Objetivo das Analises 
#========================================================


#1 - Estados com maior numero de acidentes 
#2 - Meses do ano com maior numero de ocorrencias 
#3 - Tipos de acidentes com maior frequencia 
```

```python
#===============================================================
#Carregando tabelas Silver 
#===============================================================

df_gold =spark.table("projeto_profissional_bigdata.datatran_2018.silver_datatran18")
```

```python
#=====================================================================
# Visualizar dados 
# ====================================================================

df_gold.show(5, truncate = False)
display(df_gold)
```

```python
#==============================================================
# Pergunta 01 
# Estados com maior numero de acidentes 
#==============================================================
gold_acidente_por_estado = (df_gold.groupBy("uf").agg(count("*").alias("Total_de_Acidentes")).orderBy(col("Total_de_Acidentes").desc()))
print( "Estados com mais Acidentes")
gold_acidente_por_estado.limit(27).toPandas()
```

```python
#===================================================================
#Salvar tabela Gold
#===================================================================
gold_acidente_por_estado.write\
    .format("delta")\
    .mode("overwrite")\
    .saveAsTable("projeto_profissional_bigdata.datatran_2018.gold_acidente_por_estado")
       
```

```python
%sql
select * from projeto_profissional_bigdata.datatran_2018.gold_acidente_por_estado limit (27);
```

```python
#==============================================================
#Meses do ano com maior nomero de ocorrencias
#==============================================================
gold_acidente_por_mes = df_gold.groupBy("mes_acidente","nome_mes").agg(count("*").alias("Total_Acidentes")).orderBy(col("mes_acidente"))
print("Meses com mais Acidentes")
gold_acidente_por_mes.limit(12).toPandas()
```

```python
#=================================================================
# Salvar tabela Gold
#=================================================================
gold_acidente_por_mes.write\
    .format("delta")\
    .mode("overwrite")\
    .saveAsTable("projeto_profissional_bigdata.datatran_2018.gold_acidente_por_mes")
       
```

```python
#================================
# Pergunta 3
# tipos de acidentes frequentes
#================================

gold_tipos_acidentes = (
    df_gold.groupBy("tipo_acidente")
    .agg(count("*").alias("total_acidentes"))
    .orderBy(col("total_acidentes").desc())
)

print("TIPOS DE ACIDENTES MAIS FREQUENTES")
display(gold_tipos_acidentes)
```

```python
gold_tipos_acidentes.write\
    .format("delta")\
    .mode("overwrite")\
    .saveAsTable("projeto_profissional_bigdata.datatran_2018.gold_tipo_acidentes")
```
